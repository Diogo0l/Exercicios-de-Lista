## Lista de Números

1. Crie um programa que receba uma lista de números e retorne o maior valor dessa lista.

```python
lista = [4, 8, 2, 9, 5]
maior = lista[0]
for numero in lista:
    if numero > maior:
        maior = numero
print(maior)
```

## Lista de Números com Soma

2. Crie um programa que receba uma lista de números e retorne a soma de todos os valores.

```python
lista = [4, 8, 2, 9, 4]
soma = 0
for numero in lista:
    soma += numero
print(soma)
```

## Lista com Interseção de Listas

3. Crie um programa que receba duas listas e retorne uma terceira lista com a interseção dos elementos das duas listas.

```python
lista1 = [1, 2, 3, 4, 5]
lista2 = [3, 4, 5, 6, 7]
resultado = []
for item in lista1:
    if item in lista2 and item not in resultado:
        resultado.append(item)
print(resultado)
```

## Lista de pares

4. Crie um programa que receba uma lista de números e retorne uma lista com os números pares dessa lista.

Algoritmo:

1. Leia o primeiro número e salve na variável num
2. Criar uma lista de pares
3. Descobrir se é par ou impar: dividindo por 2 e se tiver resto 0 o número é par
3. Mostrar a lista de pares.

```python
numeros = []
tamanho = int(input('Digite o tamanho da lista: '))
for i in range(tamanho):
  num = int(input('Digite um número: '))
if num % 2 == 0:
  numeros.append(num)
print(f'Números pares foram: {numeros}')
```

Desafio: Criar um algortimo para receber 3 notas e calcular a média, e mostrar o resultado com aprovado se a média for maior ou igual a 7.

Algoritmo:

1. Receber 3 notas
2. Calcular a média
3. Se a média for maior ou igual a 7 e mostrar o resultado como aprovado.
4. Senão, mostrar o resultado como reprovado.

```python
notas = []

for i in range(3):
  nota = float(input('Digite a nota: '))
  notas.append(nota)
media = sum(notas) / len(notas)
print(f'A média do aluno foi: {media}')

if media >= 7:
  print('Aprovado')
else:
  print('Reprovado')
```
